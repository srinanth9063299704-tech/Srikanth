```python
class MyThread implements Runnable {
    String msg;
    int time;

    MyThread(String m, int t) {
        msg = m;
        time = t;
    }

    public void run() {
        try {
            while(true) {
                System.out.println(msg);
                Thread.sleep(time);
            }
        } catch(Exception e) {}
    }
}

class Main {
    public static void main(String[] args) {
        new Thread(new MyThread("Good Morning", 1000)).start();
        new Thread(new MyThread("Hello", 2000)).start();
        new Thread(new MyThread("Welcome", 3000)).start();
    }
}
```
